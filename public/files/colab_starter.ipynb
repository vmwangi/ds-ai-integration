This notebook **pseudonymizes** the working copy. Note the word: hashed identifiers are pseudonymized, not anonymous, and remain personal data under the Kenya Data Protection Act and GDPR. The prompt rule still applies afterwards.

Upload `dukalink_customers.csv` to this Colab session first, then run the cell below.


```python
import hashlib

import pandas as pd

df = pd.read_csv("dukalink_customers.csv")

# Drop what the analysis does not need at all
df = df.drop(columns=["national_id"])

# Hash what must stay matchable against external records
# (payment logs, logistics partner exports) but never readable
SALT = "dukalink-2026"
df["phone_number"] = df["phone_number"].apply(
    lambda v: hashlib.sha256(f"{SALT}{v}".encode()).hexdigest()[:16])

# Your prompt context: structure and aggregates, never raw rows
df.info()
df.describe(include="all")
```
